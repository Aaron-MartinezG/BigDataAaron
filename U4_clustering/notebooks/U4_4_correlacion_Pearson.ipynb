1. Importar bibliotecas

```python
import pandas as pd
import numpy as np
```

```python
ruta = ("../data/raw/dataset_7variables_2000_2022.csv")
df = pd.read_csv(ruta)
```

4. Seleccionar variables

```python
variables_numericas = ["population",
                       "births",
                       "deaths",
                       "child_deaths_under_5",
                       "health_percentage",
                       "gdp",
                       "life_expectancy"]
```

```python
df_vn = df[variables_numericas].copy()
```

```python
from itertools import combinations
from math import comb

#Calcular el número de combinaciones
n = len(variables_numericas),
total_combinaciones = comb(n, 2)

print(f"Número de combinaciones posibles: {total_combinaciones}")
```

```python
# inicializamos el arreglo para guardar los resultados
resultados = []

for variable_x, variable_y in combinations(variables_numericas, 2):
    #se calcula r mediante la función corr de pandas
    r = df_vn[variable_x].corr(df_vn[variable_y], method="pearson")

    # se guardan en un arreglo todos los valores obtenidos
    resultados.append([variable_x, variable_y, r, abs(r)])

    # el arreglo de resultados se transforma en df de pandas para manipularlo
    #Convertir los resutltados en un DataFrame
    df_resultados = pd.DataFrame(
        resultados,
        columns=["Variable X",
                 "Variable Y",
                 "Pearson",
                 "Intensidad"] # es el valor absoluto de 4
    )

    #Ordenar por intensidad de mayor a menor
    df_resultados = df_resultados.sort_values(
        by="Intensidad",
        ascending=False
    ).reset_index(drop=True)

    display(df_resultados)
```
